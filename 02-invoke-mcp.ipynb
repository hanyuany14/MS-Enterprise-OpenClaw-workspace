##### MCP

```python
url = "https://coding-tool-contoso.lemonocean-38f621a9.eastus2.azurecontainerapps.io/mcp"

# 2. 設定 Headers
headers = {
    "Authorization": f"Bearer {USER_TOKEN}",
    "Content-Type": "application/json",
    "Accept": "application/json"
}

# 3. 設定 MCP JSON-RPC Payload (巢狀字典)
payload = {
    "jsonrpc": "2.0",
    "id": 1,
    "method": "tools/call",
    "params": {
        "name": "run_coding_workflow",
        "arguments": {
            "request": user_requests,
            "session_id": session_id
        }
    }
}

try:
    # 3. 發送 POST 請求
    response = requests.post(url, headers=headers, json=payload)
    response.raise_for_status()
    
    # 4. 第一層解析：取得 MCP 外殼 (Envelope)
    mcp_envelope = response.json()
    
    # 5. 第二層解析：深入 MCP 的 content 陣列並還原內部的 JSON 字串
    # 結構路徑: result -> content -> [0] -> text
    mcp_content = mcp_envelope.get("result", {}).get("content", [])
    
    if mcp_content and "text" in mcp_content[0]:
        stringified_json = mcp_content[0]["text"]
        
        # 關鍵步驟：將字串轉回字典物件
        actual_data = json.loads(stringified_json)
        
        # 6. 漂亮列印最終資料
        print("=== 成功解析 MCP 內部資料 ===")
        print(json.dumps(actual_data, indent=4, ensure_ascii=False))
        
        # 您可以像這樣存取欄位：
        # print(f"任務狀態: {actual_data.get('status')}")
        
    else:
        print("警告：MCP 回應中找不到 text 內容。")
        print("原始回傳：", mcp_envelope)

except requests.exceptions.RequestException as e:
    print(f"網路或 API 請求錯誤: {e}")
except json.JSONDecodeError as e:
    print(f"內部資料解析失敗 (不是有效的 JSON 字串): {e}")
except Exception as e:
    print(f"發生未預期錯誤: {e}")
```
